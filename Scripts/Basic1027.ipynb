```python
item1, price1 = "Americano", 4500
item2, price2 = "Bage1", 3800
item3, price3 = "Orange Juice", 5200

total = price1 + price2 + price3

print("=" * 26)
print(f"{'ITEM':<14}{'PRICE':>12}")
print(f"{item1:<14}{price1:>12}")
print(f"{item2:<14}{price2:>12}")
print(f"{item3:<14}{price3:>12}")
print("-" * 26)
print(f"{'TOTAL':<14}{total:>12,}")
```

```python
total_sec = int(input("초: "))

hours = total_sec // 3600
minutes = total_sec % 3600 // 60
seconds = total_sec % 60

print(f"{hours}시간 {minutes}분 {seconds}초")
print(f"{hours:02d}:{minutes:02d}:{seconds:02d}")
```
