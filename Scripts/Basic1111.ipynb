```python
def func_a(numbers):
    return sum(numbers)

def func_b(numbers):
    return len(numbers)

def solution(numbers):
    total = func_a(numbers)
    count = func_b(numbers)
    return total // count

print(solution([10, 20, 30]))
print(solution([1, 2]))
print(solution([7]))
```

```python
def func_a(numbers):
    total = 0
    for x in numbers:
        total += x
    return total

def func_b(numbers):
    count = 0
    for _ in numbers:
        count+=1
    return count

def solution(numbers):
    total = func_a(numbers)
    count = func_b(numbers)
    return total // count

print(solution([10, 20, 30]))
print(solution([1, 2]))
print(solution([7]))
```
