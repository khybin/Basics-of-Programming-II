```python
def solution(socres):
    s = list(map(int, socres.split()))
    print(f"{len(s)}명, 평균 {sum(s) / len(s):.1f}점")
    print(f"점수 차이: {max(s) - min(s)}점")

solution("85 92 78 64 90")
```

```python
def solution(numbers):
    answer = []
    for x in numbers:
        if x > 0:
            answer.append(x)
    return answer

print(solution([3, -1, 0, 5, -7]))
print(solution([-2, -4]))
```

```python
# 예: 일주일 동안의 걸음 수를 리스트에 담고, 평균과 가장 많이 걸은 날을 출력해 보세요.
def solution(walks:list):
    days = ['월', '화', '수', '목', '금', '토', '일']
    index = walks.index(max(walks))
    day = days[index]
    avg = sum(walks) / len(walks)
    avg = int(avg) if avg.is_integer() else round(avg, 1)
    print(f"일주일 동안의 걸음 수의 평균은 {avg}이고, 가장 많이 걸은 날은 {day}요일입니다.")

solution([3000, 5000, 7000, 2000, 9000, 4000, 6000])
```
