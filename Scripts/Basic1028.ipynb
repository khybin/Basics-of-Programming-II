```python
kor, eng, mat = 72, 55, 88
avg = (kor + eng + mat) / 3
no_fail = kor >= 40 and eng >= 40 and mat >= 40
passed = avg >= 60 and no_fail
print(f"{avg:.1f} {passed}")
```

```python
year = int(input("몇 년도입니까?"))

if year % 4 == 0 and year % 100 != 0 or year % 400 == 0:
    print(f"{year}년은 윤년입니다!")
else:
    print(f"{year}년은 윤년이 아니예요!")
```
